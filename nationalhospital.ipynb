```python
## data reading
patients_df = spark.read.csv(
    f"{file_path}/raw/patients_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)

imaging_df = spark.read.csv(
    f"{file_path}/raw/imaging_results_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)

lab_df = spark.read.csv(
    f"{file_path}/raw/lab_results_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)

med_records_df = spark.read.csv(
    f"{file_path}/raw/medical_records_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)

trials_df = spark.read.csv(
    f"{file_path}/raw/clinical_trials_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)

participants_df = spark.read.csv(
    f"{file_path}/raw/trial_participants_data.csv",
    header=True,
    inferSchema=True,
    quote='"',
    escape='"',
    multiLine=True
)
```

```python
## Cleaning and transformation
from pyspark.sql.functions import *

## Patients data
processed_patients_df = patients_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date()) \
    .write.mode("overwrite").csv(
        f"{file_path}/processed_data/patients_data_processed",
        header=True
    )

## Imaging data
processed_imaging_df = imaging_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date())

processed_imaging_df.write.mode("overwrite").csv(
    f"{file_path}/processed_data/imaging_processed",
    header=True
)

## Lab data
processed_lab_df = lab_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date())

processed_lab_df.write.mode("overwrite").csv(
    f"{file_path}/processed_data/lab_processed",
    header=True
)

## Medical records data
processed_med_records_df = med_records_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date())

processed_med_records_df.write.mode("overwrite").csv(
    f"{file_path}/processed_data/med_records_processed",
    header=True
)

## Clinical trials data
processed_trials_df = trials_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date())

processed_trials_df.write.mode("overwrite").csv(
    f"{file_path}/processed_data/trials_processed",
    header=True
)

## Trial participants data
processed_participants_df = participants_df.dropDuplicates() \
    .na.drop() \
    .withColumn("loaded_date", current_date())

processed_participants_df.write.mode("overwrite").csv(
    f"{file_path}/processed_data/participants_processed",
    header=True
)
```
